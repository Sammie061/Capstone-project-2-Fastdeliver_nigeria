```python
pos_df = pos.copy()
```

```python
# 1. First convert to datetime - do this BEFORE filling
pos_df['order_date'] = pd.to_datetime(pos_df['order_date'], errors='coerce')
pos_df['pickup_time'] = pd.to_datetime(pos_df['pickup_time'], errors='coerce')
pos_df['delivery_time'] = pd.to_datetime(pos_df['delivery_time'], errors='coerce')
pos_df['order_value'] = pd.to_numeric(pos_df['order_value'], errors='coerce')

# 3. Check dtypes now
print(pos_df[['order_date','pickup_time','delivery_time','order_value']].dtypes)
print(pos_df.isnull().sum())
```

```python
#fill missing values/blanks with mean for order_value and rider_rating
#.1 Fill numeric columns with mean
pos_df['pickup_time'] = pos_df['pickup_time'].fillna(pos_df['pickup_time'].mean())
pos_df['delivery_time'] = pos_df['delivery_time'].fillna(pos_df['delivery_time'].mean())
pos_df['rider_rating'] = pos_df['rider_rating'].fillna(pos_df['rider_rating'].mean())

#.2 Fill categorical with mode
pos_df['payment_method'] = pos_df['payment_method'].fillna(pos_df['payment_method'].mode()[0])

## Dropping missing values
# For pickup_time & delivery_time - DO NOT FILL WITH MEAN
# Just drop rows where time is missing - best for delivery analysis
#pos_df = pos_df.dropna(subset=['pickup_time', 'delivery_time', 'order_value'])
pos_df = pos_df.dropna(subset=['order_value'])
pos_df.isnull().sum()
```

```python
## Dropping duplicates row
pos_df.drop_duplicates(inplace=True)
pos_df.duplicated().sum()
```

```python
pos_df
```

```python
# Now try - find the zone column
zone_col = [c for c in pos_df.columns if 'zone' in c or 'area' in c or 'location' in c][0]
print(f"Using '{zone_col}' as zone column")
```

```python
### At least 2 groupby analyses 
# Groupby 1: Revenue by zone
revenue_by_zone = pos_df.groupby(zone_col)['order_value'].agg(['sum','mean','count']).round(2)
revenue_by_zone = revenue_by_zone.sort_values('sum', ascending=False)
print(revenue_by_zone.head())
```

```python
# Groupby 2: by rider
rider_col = [c for c in pos_df.columns if 'rider' in c.lower()][0]

rider_performance = pos_df.groupby(rider_col).agg(
    total_orders=('order_id','count'),
    avg_rating=('rider_rating','mean'),
    total_revenue=('order_value','sum')
).round(2).sort_values('total_revenue', ascending=False)

print(rider_performance.head())
```

```python
# Groupby 3: by payment method
pay_col = [c for c in pos_df.columns if 'payment' in c][0]
payment_analysis = pos_df.groupby(pay_col)['order_value'].mean().round(2)
print(payment_analysis)
```

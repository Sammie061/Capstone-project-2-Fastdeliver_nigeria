##  Section 2 — Logical Operators

Comparisons answer one question at a time. **Logical operators combine multiple questions** into a single answer.

| Operator | Rule | Example |
|---|---|---|
| `and` | Both conditions must be `True` | Delayed AND overcrowded |
| `or` | At least one condition must be `True` | Delayed OR speeding |
| `not` | Flips `True` to `False` and vice versa | NOT payment confirmed |

---
### 2.1  `and` — both must be true

````python
# ── and: both conditions True ───────────
# Business question: Which trips are BOTH delayed AND overcrowded (>80% load)?

trip = schedule[2]   # MON-EV: R310, speed=95, pax=61/60
print("Trip:", trip["day"], trip["period"], trip["route"])
print()

is_delayed   = trip["is_delayed"]
load_factor  = trip["passengers"] / 60
is_overcrowded = load_factor > 0.80

print(f"  is_delayed:     {is_delayed}")
print(f"  load_factor:    {load_factor:.2f}  ({load_factor*100:.0f}%)")
print(f"  is_overcrowded: {is_overcrowded}")
print()
print(f"  Delayed AND overcrowded: {is_delayed and is_overcrowded}")
print()

# Real-world application
print("── Checking all 21 trips for 'delayed AND overcrowded' ──")
count = 0
for t in schedule:
    if t["is_delayed"] and (t["passengers"] / 60 > 0.80):
        count += 1
        print(f"  {t['day']}-{t['period']}  {t['route']}  "
              f"pax={t['passengers']}/60  delayed={t['is_delayed']}")
print(f"Total: {count} trips need priority dispatch")
````

---
### 2.2  `or` — at least one must be true

````python
# ── or: either condition True ───────────────────────
# Business question: Which trips have ANY safety concern?
# (speeding OR critically low fuel OR overloaded)

print("── Safety concern trips (speeding OR low fuel OR overloaded) ──")
safety_concerns = []

for t in schedule:
    is_speeding    = t["speed_kmh"] > 85
    is_low_fuel    = t["fuel_pct"]  < 35
    is_overloaded  = t["passengers"] > 60

    # or: flag this trip if ANY of the three conditions are true
    if is_speeding or is_low_fuel or is_overloaded:
        flags = []
        if is_speeding:  flags.append(f"speed={t['speed_kmh']}")
        if is_low_fuel:  flags.append(f"fuel={t['fuel_pct']}%")
        if is_overloaded:flags.append(f"pax={t['passengers']}")
        safety_concerns.append(t)
        print(f"   {t['day']}-{t['period']}  {t['route']}  →  {', '.join(flags)}")

print()
print(f"{len(safety_concerns)} trips flagged for safety review out of {len(schedule)}")
````

---
### 2.3  `not` — flip the result

````python
# ── not: reverse the boolean ──────────────────────────────────────────────────
# Business question: Show me only the trips that are running perfectly on time

print("── On-time trips (not delayed) ──")
for t in schedule:
    if not t["is_delayed"]:      # same as: t["is_delayed"] == False
        print(f"  {t['day']}-{t['period']}  {t['route']}  "
              f"speed={t['speed_kmh']} km/h")

print()

# not with complex conditions
print("── Trips that are NOT safety-critical ──")
safe_count = 0
for t in schedule:
    is_safe = not (t["speed_kmh"] > 85 or t["fuel_pct"] < 35 or t["passengers"] > 60)
    if is_safe:
        safe_count += 1

print(f"{safe_count} trips running without any safety flag")
````

````python
# ── DE insight: short-circuit evaluation ───────────
# Python stops evaluating 'and' as soon as it finds a False
# Python stops evaluating 'or'  as soon as it finds a True
# This matters for performance with large datasets

# Safe example: check type before comparing
value = "350.50"   # imagine a badly-typed fare from a legacy record

# WRONG: would raise TypeError
# if value > 300: ...

# RIGHT: check type first — if isinstance fails, the > never runs
if isinstance(value, (int, float)) and value > 300:
    print("Premium fare")
else:
    print(f"Cannot compare — value is {type(value).__name__}: {value!r}")

print()
print("Short-circuit saved us from a TypeError!")
print("This is why you always validate types before comparing in a DE pipeline.")
````

### 3.2  Build `classify_trip()` — the RAG classification function

This is the centrepiece of Class 3. We'll build the function that the CEO needs — one that reads a trip and returns **RED**, **AMBER**, or **GREEN** based on multiple factors.

````python
# ── classify_trip() — built step by step ─────────────────────────────────────

def classify_trip(trip):
    '''
    Classify a LagosGrid trip as RED, AMBER, or GREEN.
    
    RED   = immediate action required
    AMBER = monitor closely
    GREEN = operating normally
    
    Parameters
    ----------
    trip : dict
        A single trip record from the LagosGrid schedule.
    
    Returns
    -------
    str : "RED", "AMBER", or "GREEN"
    '''
    speed       = trip["speed_kmh"]
    load_factor = trip["passengers"] / 60     # 60 = bus capacity
    delayed     = trip["is_delayed"]
    fuel        = trip["fuel_pct"]

    # ── RED: immediate action ───────────
    # Speed > 90 km/h is a safety emergency — always RED
    # Delayed AND overcrowded (>90%) = operational crisis
    if speed > 90 or (delayed and load_factor > 0.90):
        return "RED"

    # ── AMBER: monitor closely ───────────
    # Delayed for any reason, OR load is high, OR fuel is getting low
    elif delayed or load_factor > 0.80 or fuel < 35:
        return "AMBER"

    # ── GREEN: all good ───────────────────────────────────────────────────────
    else:
        return "GREEN"


# ── Test on individual trips ──────────────────────────────────────────────────
test_trips = [schedule[0], schedule[1], schedule[2], schedule[3], schedule[4]]

print("Testing classify_trip() on 5 trips:")
print()
for t in test_trips:
    status = classify_trip(t)
    icon   = "Red-ICON" if status == "RED" else "AMBER-ICON" if status == "AMBER" else "GREEN-ICON"
    print(f"  {icon} {t['day']}-{t['period']}  {t['route']:<6}  "
          f"speed={t['speed_kmh']:4.1f}  "
          f"load={t['passengers']/60*100:.0f}%  "
          f"delayed={str(t['is_delayed']):<5}  "
          f"fuel={t['fuel_pct']:2d}%  - {status}")
````

### 4.6  Mini Challenge — Build the route stats dictionary

````python
# ── CHALLENGE: Build per-route statistics using a for loop ────────────────────
# Loop through all 21 trips and accumulate stats per route.
# The result should look like:
#
# {
#   "R101": {"trips": 7, "delays": 0, "total_fare": 2240.0, "fares": [...]},
#   "R207": {"trips": 7, "delays": 4, "total_fare": ...,    "fares": [...]},
#   "R310": {"trips": 7, "delays": 2, "total_fare": ...,    "fares": [...]},
# }

route_stats = {}

for trip in schedule:
    r = trip["route"]

    # Initialise the route entry if we haven't seen it yet
    if r not in route_stats:
        route_stats[r] = {"trips": 0, "delays": 0, "total_fare": 0.0,
                          "total_pax": 0, "fares": [], "statuses": []}

    # Accumulate values
    route_stats[r]["trips"]      += 1
    route_stats[r]["delays"]     += int(trip["is_delayed"])   # True=1, False=0
    route_stats[r]["total_fare"] += trip["fare"]
    route_stats[r]["total_pax"]  += trip["passengers"]
    route_stats[r]["fares"].append(trip["fare"])
    route_stats[r]["statuses"].append(classify_trip(trip))

# ── Print the accumulated stats ───────────────────────────────────────────────
print("── Route Statistics (from loop) ────────────────────────────────────────")
print()
for route, stats in route_stats.items():
    avg_fare  = stats["total_fare"] / stats["trips"]
    avg_pax   = stats["total_pax"]  / stats["trips"]
    delay_pct = stats["delays"]     / stats["trips"] * 100
    red_count = stats["statuses"].count("RED")

    print(f"  {route}")
    print(f"    Trips:      {stats['trips']}")
    print(f"    Delays:     {stats['delays']}  ({delay_pct:.0f}%)")
    print(f"    Avg fare:  ₦{avg_fare:.2f}")
    print(f"    Avg pax:    {avg_pax:.1f}/60  ({avg_pax/60*100:.0f}% load)")
    print(f"    RED alerts: {red_count}")
    print()

print("route_stats is now ready for the weekly report in Section 7!")
````

##  Section 5 — while loops

A `while` loop **keeps running as long as a condition is True**. Unlike a `for` loop, you control when it stops.

```python
while condition:
    do_something()
    # update something so condition eventually becomes False
```

In DE, `while` loops appear most in:
- **Retry logic** — keep trying a DB connection until it works or max attempts reached
- **Streaming** — keep reading events while the queue has data
- **Simulation** — run until a capacity or threshold is hit

### 5.1  Bus boarding simulation

````python
# ── Simulate passengers boarding a bus until it's full ────────────────────────
import random
random.seed(42)   # fixed seed so results are reproducible

capacity    = 60
seats_taken = 0
boarding_log = []

print("── Bus R101 Boarding Simulation ───────────────────")
print()

while seats_taken < capacity:
    # Each boarding event: 1-3 passengers get on at each stop
    boarding = random.randint(1, 3)

    # Don't exceed capacity
    boarding = min(boarding, capacity - seats_taken)
    seats_taken += boarding
    load_pct = seats_taken / capacity * 100

    boarding_log.append(seats_taken)

    status = "FULL" if seats_taken == capacity else f"{load_pct:.0f}% load"
    print(f"  {len(boarding_log):2d} stops  +{boarding} passengers  →  {seats_taken:2d}/{capacity}  [{status}]")

    if seats_taken >= capacity:
        print()
        print("  Bus is full — doors closed, no more boarding.")
        break

print()
print(f"Bus filled after {len(boarding_log)} stops.")
print(f"Boarding log: {boarding_log}")
````

### 5.2  Pipeline retry pattern — the most important while loop in DE

````python
# ── Retry pattern: connect to a data source with exponential backoff ───────────
import time

def attempt_db_connection(attempt_num):
    '''Simulates a DB connection — fails first 2 attempts, succeeds on 3rd.'''
    # In real code this would be: psycopg2.connect(...) or sqlalchemy.create_engine(...)
    if attempt_num < 3:
        raise ConnectionError(f"Database unreachable (attempt {attempt_num})")
    return "CONNECTION_OK"


MAX_ATTEMPTS = 5
attempt      = 0
connected    = False
wait_seconds = 1    # start with 1 second wait, double each time

print("── Connecting to LagosGrid database ────────────────────────────────────")
print()

while attempt < MAX_ATTEMPTS and not connected:
    attempt += 1
    print(f"  Attempt {attempt}/{MAX_ATTEMPTS}...", end=" ")

    try:
        result = attempt_db_connection(attempt)
        connected = True
        print(f" {result}")
    except ConnectionError as e:
        print(f" {e}")
        if attempt < MAX_ATTEMPTS:
            print(f"    Retrying in {wait_seconds}s...")
            # time.sleep(wait_seconds)   # uncomment in real code
            wait_seconds *= 2            # exponential backoff: 1s - 2s - 4s - 8s

print()
if connected:
    print(" Pipeline connected — beginning data ingestion.")
else:
    print(" Max retries exceeded — alert the on-call engineer.")
````

## Section 6 — List Comprehensions

A **list comprehension** is a concise way to build a list from another sequence — combining a `for` loop and optional filtering into a single expression.

```python
# Full syntax
new_list = [expression  for item in sequence  if condition]

#   what to keep       the loop             optional filter
```

They're everywhere in production DE code. Once you're comfortable with for loops, comprehensions are the natural upgrade.

---
### 6.1  The collapse — from loop to comprehension

````python
# ── The same task: two ways ─────────────────────────────

# ── WAY 1: Traditional for loop ───────────────────────────────────────────────
delayed_trips_loop = []

for trip in schedule:
    if trip["is_delayed"]:
        delayed_trips_loop.append(trip)

print(f"For loop:        {len(delayed_trips_loop)} delayed trips")

# ── WAY 2: List comprehension ─────────────────────────────────────────────────
delayed_trips_comp = [trip for trip in schedule if trip["is_delayed"]]

print(f"Comprehension:   {len(delayed_trips_comp)} delayed trips")
print(f"Same result?     {delayed_trips_loop == delayed_trips_comp}")
print()
print("Read the comprehension as:")
print("  [  trip                      -  keep the full trip dict")
print("     for trip in schedule      -  from every trip in the schedule")
print("     if trip['is_delayed']  ]  -  but only if it's delayed")
````

---
### 6.2  Transform + filter

````python
# ── Comprehensions that transform as well as filter ────────────────────────────

# Extract just the fare values from all trips
all_fares = [trip["fare"] for trip in schedule]
print(f"All fares: {all_fares}")
print(f"  min=₦{min(all_fares):.2f}  max=₦{max(all_fares):.2f}  avg=₦{sum(all_fares)/len(all_fares):.2f}")
print()

# Fares above ₦340 only (filter + extract)
premium_fares = [trip["fare"] for trip in schedule if trip["fare"] > 340]
print(f"Premium fares (>₦340): {premium_fares}")
print()

# Build trip summary strings (transform)
summaries = [
    f"{t['day']}-{t['period']}  {t['route']}  {classify_trip(t)}"
    for t in schedule
]
print("Trip summaries (first 5):")
for s in summaries[:5]:
    print(f"  {s}")
print()

# Classify ALL trips in one line
all_statuses = [classify_trip(t) for t in schedule]
print(f"All statuses: {all_statuses}")
print(f"RED: {all_statuses.count('RED')}  AMBER: {all_statuses.count('AMBER')}  GREEN: {all_statuses.count('GREEN')}")
````

## Section 7 — Mini Project: LagosGrid Week 1 Performance Report

**Brief:** The CEO needs the Week 1 performance report before Monday's board meeting.
Build a complete Python analytics script that processes all 21 trips and prints a formatted summary.

**Deliverable:** A formatted report with per-route statistics, RAG status, and a summary footer.

---
### Step 1 — Classify every trip and collect RED alerts

````python
# ── STEP 1: Run classification across all 21 trips ────────────────────────────

trip_classifications = []
red_alerts           = []

for i, trip in enumerate(schedule, start=1):
    status = classify_trip(trip)
    trip_classifications.append({**trip, "trip_num": i, "status": status})

    if status == "RED":
        red_alerts.append({"trip_num": i, **trip, "status": status})

print(f"Classification complete: {len(trip_classifications)} trips processed")
print(f"RED alerts: {len(red_alerts)}")
print()

# Preview the RED alerts
if red_alerts:
    print("RED trips requiring immediate attention:")
    for alert in red_alerts:
        reason = []
        if alert["speed_kmh"] > 90:       reason.append(f"speed={alert['speed_kmh']} km/h")
        if alert["is_delayed"] and alert["passengers"]/60 > 0.90:
            reason.append(f"delayed+overcrowded ({alert['passengers']}/60)")
        print(f"  Trip {alert['trip_num']:02d}  {alert['day']}-{alert['period']}  "
              f"{alert['route']}  →  {', '.join(reason)}")
````

---
### Step 2 — Aggregate per-route statistics

````python
# ── STEP 2: Build route_stats (building on the challenge from Section 4) ───────

route_stats = {}

for tc in trip_classifications:
    r = tc["route"]
    if r not in route_stats:
        route_stats[r] = {
            "trips": 0, "delays": 0, "total_fare": 0.0,
            "total_pax": 0, "max_speed": 0.0, "min_fuel": 100,
            "red": 0, "amber": 0, "green": 0,
        }

    route_stats[r]["trips"]      += 1
    route_stats[r]["delays"]     += int(tc["is_delayed"])
    route_stats[r]["total_fare"] += tc["fare"]
    route_stats[r]["total_pax"]  += tc["passengers"]
    route_stats[r]["max_speed"]   = max(route_stats[r]["max_speed"], tc["speed_kmh"])
    route_stats[r]["min_fuel"]    = min(route_stats[r]["min_fuel"],  tc["fuel_pct"])
    route_stats[r][tc["status"].lower()] += 1

print("Per-route aggregation complete:")
for route, stats in sorted(route_stats.items()):
    print(f"  {route}: {stats['trips']} trips  "
          f"delays={stats['delays']}  "
          f"avg_fare=₦{stats['total_fare']/stats['trips']:.0f}  "
          f"RED={stats['red']}  AMBER={stats['amber']}  GREEN={stats['green']}")
````

---
### Step 3 — Assign overall RAG status per route

````python
# ── STEP 3: Determine each route's overall weekly status ──────────────────────

def get_route_rag(stats):
    '''
    Assign overall weekly RAG status to a route.
    
    RED   = any RED trips, OR delay rate > 40%
    AMBER = delay rate > 20%, OR any AMBER trips, OR fuel dropped below 30
    GREEN = otherwise
    '''
    trips      = stats["trips"]
    delay_rate = stats["delays"] / trips

    if stats["red"] > 0 or delay_rate > 0.40:
        return "RED"
    elif delay_rate > 0.20 or stats["amber"] > 0 or stats["min_fuel"] < 30:
        return "AMBER"
    else:
        return "GREEN"


route_rag = {route: get_route_rag(stats) for route, stats in route_stats.items()}

print("Route overall RAG status:")
for route, rag in sorted(route_rag.items()):
    icon = "RED-ALERT" if rag == "RED" else "AMBER-ALERT" if rag == "AMBER" else "GREEN"
    print(f"  {icon}  {route}  →  {rag}")
````

````python
# ── STEP 4: Generate the Week 1 Performance Report ────────────────────────────

sep_thick = "═" * 60
sep_thin  = "─" * 60

# ── Header ────────────────────────
print(sep_thick)
print(f"{'LAGOSGRID — WEEK 1 PERFORMANCE REPORT':^60}")
print(f"{'Generated: Saturday, July 4 2026':^60}")
print(sep_thick)

# ── Route table ───────────────────────────────────────────────────────────────
header = f"  {'Route':<8} {'Trips':>5} {'Delays':>7} {'Del%':>6} {'Avg Fare':>9} {'Load':>6} {'Peak Spd':>9} {'RAG':>6}"
print()
print(header)
print("  " + sep_thin)

for route in sorted(route_stats.keys()):
    s    = route_stats[route]
    rag  = route_rag[route]
    icon = "RED-ALERT" if rag == "RED" else "AMBER-ALERT" if rag == "AMBER" else "GREEN"

    avg_fare   = s["total_fare"] / s["trips"]
    avg_load   = s["total_pax"]  / s["trips"] / 60 * 100
    delay_pct  = s["delays"]     / s["trips"] * 100

    print(f"  {route:<8} {s['trips']:>5} {s['delays']:>7} "
          f"{delay_pct:>5.0f}% {avg_fare:>8.2f}  {avg_load:>4.0f}%  "
          f"{s['max_speed']:>7.1f}   {icon} {rag}")

# ── Alert summary ──────────────────────────────────────────────────────────────
print()
print("  " + sep_thin)
total_red   = sum(s["red"]   for s in route_stats.values())
total_amber = sum(s["amber"] for s in route_stats.values())
total_green = sum(s["green"] for s in route_stats.values())
total_delay = sum(s["delays"] for s in route_stats.values())

print()
print(f"  FLEET SUMMARY:")
print(f"  Total trips this week:  {len(schedule)}")
print(f"  Delayed trips:          {total_delay}  ({total_delay/len(schedule)*100:.0f}%)")
print(f"  = RED trips:            {total_red}")
print(f"  = AMBER trips:          {total_amber}")
print(f"  = GREEN trips:          {total_green}")
print(f"  Total revenue:         ₦{sum(s['total_fare'] for s in route_stats.values()):,.2f}")

# ── RED alerts detail ─────────────────────────────────────────────────────────
if red_alerts:
    print()
    print("   RED ALERTS — IMMEDIATE ACTION REQUIRED")
    for alert in red_alerts:
        print(f"  Trip {alert['trip_num']:02d}  {alert['day']}-{alert['period']}  "
              f"{alert['route']}  Bus {alert['bus_id']}  "
              f"speed={alert['speed_kmh']:.1f} km/h  pax={alert['passengers']}/60")

# ── Footer ────────────────────────────────────────────────────────────────────
print()
print(sep_thick)
print(f"{'Report prepared by: LagosGrid Analytics Pipeline':^60}")
print(f"{'Python 3  ·  10Alytics DE Programme  ·  Class 3':^60}")
print(sep_thick)
````

##  Bonus: Refactor into Functions

Good Python code doesn't live in one long script — it's broken into **reusable functions**.
This is what your capstone project should look like structurally.

Here's the Week 1 report refactored so every step is its own function:

````python
# ── All logic refactored into named functions ──────────────────────────────────

def classify_trip(trip):
    '''Return RED / AMBER / GREEN for a single trip dict.'''
    speed = trip["speed_kmh"]
    load  = trip["passengers"] / 60
    delay = trip["is_delayed"]
    fuel  = trip["fuel_pct"]
    if speed > 90 or (delay and load > 0.90):  return "RED"
    elif delay or load > 0.80 or fuel < 35:    return "AMBER"
    else:                                       return "GREEN"


def aggregate_routes(schedule):
    '''Build per-route statistics dict from a list of trip records.'''
    stats = {}
    for t in schedule:
        r = t["route"]
        if r not in stats:
            stats[r] = {"trips":0,"delays":0,"total_fare":0.0,
                        "total_pax":0,"max_speed":0.0,"min_fuel":100,
                        "red":0,"amber":0,"green":0}
        s = stats[r]
        s["trips"]      += 1
        s["delays"]     += int(t["is_delayed"])
        s["total_fare"] += t["fare"]
        s["total_pax"]  += t["passengers"]
        s["max_speed"]   = max(s["max_speed"], t["speed_kmh"])
        s["min_fuel"]    = min(s["min_fuel"],  t["fuel_pct"])
        status           = classify_trip(t)
        s[status.lower()] += 1
    return stats


def get_route_rag(stats):
    '''Return overall RAG for a route from its aggregated stats.'''
    dr = stats["delays"] / stats["trips"]
    if stats["red"] > 0 or dr > 0.40:         return "RED"
    elif dr > 0.20 or stats["amber"] > 0 or stats["min_fuel"] < 30:
                                               return "AMBER"
    else:                                      return "GREEN"


def print_weekly_report(schedule):
    '''End-to-end: classify, aggregate, and print the CEO report.'''
    stats     = aggregate_routes(schedule)
    route_rag = {r: get_route_rag(s) for r, s in stats.items()}
    red_trips = [t for t in schedule if classify_trip(t) == "RED"]

    sep = "═" * 55
    print(sep)
    print(f"{'LAGOSGRID — WEEK 1 REPORT':^55}")
    print(sep)
    print(f"  {'Route':<7} {'Trips':>5} {'Del%':>5} {'AvgFare':>8} {'RAG'}")
    print("  " + "─" * 40)
    for r in sorted(stats):
        s  = stats[r]
        icon = "RED-ALERT" if route_rag[r]=="RED" else "AMBER-ALERT" if route_rag[r]=="AMBER" else "GREEN"
        print(f"  {r:<7} {s['trips']:>5} {s['delays']/s['trips']*100:>4.0f}%  "
              f"₦{s['total_fare']/s['trips']:>7.2f}  {icon} {route_rag[r]}")
    print("  " + "─" * 40)
    print(f"  Trips: {len(schedule)}  |  "
          f"RED: {sum(s['red']   for s in stats.values())}  "
          f"AMBER: {sum(s['amber'] for s in stats.values())}  "
          f"GREEN: {sum(s['green'] for s in stats.values())}")
    print(sep)


# ── Run it with one function call ─────────────────────────────────────────────
print_weekly_report(schedule)
````

## Section 1 : What is a Python Data Type?

A **data type** tells Python:
1. What **kind of value** a variable holds
2. What **operations are legal** on it
3. How much **memory** to allocate

Python has **8 core types** you will use constantly as a DE:

| Category | Types |
|---|---|
| **Scalars** (single values) | `int` · `float` · `str` · `bool` |
| **Collections** (multiple values) | `list` · `tuple` · `dict` · `set` |

You can always ask Python what type something is:

```python
# ── Checking types with type() ────

print(type(1042))          # int
print(type(350.50))        # float
print(type("Mile 2"))      # str
print(type(True))          # bool
print(type([1, 2, 3]))     # list
print(type((6.52, 3.37)))  # tuple
print(type({"key": "val"}))# dict
print(type({"R1", "R2"}))  # set
```

### 2.1  `int` — Integer (whole numbers)

Integers are **whole numbers** with no decimal point.  
At LagosGrid, we use `int` for anything we *count* or *identify*.

```python
# ── int examples at LagosGrid ──────

bus_id          = 1042    # unique bus identifier
passenger_count = 45      # riders on this trip
stops_count     = 12      # number of stops on the route
route_code      = 207     # numeric route ID

print("Bus ID:           ", bus_id)
print("Passengers:       ", passenger_count)
print("Stops:            ", stops_count)
print("Route code:       ", route_code)
print()

# ── Int arithmetic ────────────────────────────────────────────────────────────
total_riders   = 120 + 45           # addition
delay_minutes  = 37 - 25            # subtraction
trips_per_hour = 96 // 8            # floor division  → 12 (no remainder)
remainder      = passenger_count % 10  # modulo → last digit

print("Total riders:     ", total_riders)
print("Delay (mins):     ", delay_minutes)
print("Trips/hour:       ", trips_per_hour)
print("Count mod 10:     ", remainder)
print("Type of bus_id:   ", type(bus_id))
```

---
### 2.2  `float` — Floating-point 

Floats are numbers **with a decimal point**.  
At LagosGrid, we use `float` for fares, distances, and GPS coordinates.

```python
# ── float examples at LagosGrid ───────────

fare_amount  = 350.50    # passenger fare in Naira
distance_km  = 14.7      # route distance in kilometres
latitude     = 6.5244    # GPS latitude  
longitude    = 3.3792    # GPS longitude
speed_kmh    = 62.3      # current bus speed

print("Fare:          ", fare_amount)
print("Distance:       ", distance_km, "km")
print("GPS:            ", latitude, ",", longitude)
print("Speed:          ", speed_kmh, "km/h")
print()

# ── Float arithmetic ────────
total_fare    = fare_amount * 3          # 3 trips
avg_fare      = total_fare / 3           # division - always float
fare_rounded  = round(avg_fare, 2)       # 2 decimal places

print("3-trip total:  ", total_fare)
print("Average fare:  ", avg_fare)
print("Rounded:       ", fare_rounded)
print()

# ── Important: int vs float division ────────────
print("10 / 3  =", 10 / 3)     # - 3.333...  (float)
print("10 // 3 =", 10 // 3)    # - 3          (int — floor division)
print("int(3.9)=", int(3.9))   # - 3          (truncates, does NOT round)
```

> **GPS precision matters.** `6.5244` and `6.5` are about 2.7 km apart.  
> Always store coordinates as `float` with full precision — never round them.

### 2.3  `str` — String (text)

Strings are **text values enclosed in quotes** (single or double).  
At LagosGrid, we use `str` for names, codes, statuses, and timestamps.

```python
# ── str examples at LagosGrid ──────────────

route_name     = "Mile 2 - TBS Express"
driver_name    = "Chukwuemeka Obi"
status         = "ON_ROUTE"
departure_time = "07:30:00"    # looks like a time, but IS A STRING until parsed
vehicle_type   = 'BRT'         # single quotes work too

print("Route:      ", route_name)
print("Driver:     ", driver_name)
print("Status:     ", status)
print("Departure:  ", departure_time)
print("Type of departure_time:", type(departure_time))  # - str, not a time object
print()

# ── f-strings — the DE way to format output ───────────────────────────────────
print(f"Bus {vehicle_type} on route '{route_name}' — Status: {status}")
print(f"Driver {driver_name.split()[0]} departing at {departure_time}")
print()

# ── Useful string methods for DE ──────────────────────────────────────────────
print(status.upper())                         # - 'ON_ROUTE'
print(driver_name.lower())                    # - 'chukwuemeka obi'
print(route_name.split(" - "))                # - ['Mile 2 ', ' TBS Express']
print(driver_name.replace(" ", "_"))          # - 'Chukwuemeka_Obi'
print("  ON_ROUTE  ".strip())                 # - 'ON_ROUTE'  (removes whitespace)
print("Mile 2 Express".startswith("Mile"))    # - True
```

> **DE Warning:** `"07:30:00"` is a string. It looks like a time, but you cannot do
> `departure_time + 30` to get `08:00:00`. You must first parse it into a proper time object.
> This is one of the most common traps when reading CSV files or API responses.

##  Section 3 — Collection Types

Scalars store **one value**. Collections store **many values**.  
Every real-world DE system is built on collections — they are the data structures you will use every single day.

| Collection | Brackets | Ordered | Mutable | Duplicates |
|---|---|---|---|---|
| `list`  | `[ ]` |  Yes |  Yes |  Yes |
| `tuple` | `( )` |  Yes |  No  |  Yes |
| `dict`  | `{ }` with colons |  Yes | Yes |  No keys |

---
### 3.1  `list` — Ordered, Mutable Sequence

Use a `list` when:
- The **order matters** (you need position 0, 1, 2…)
- The **data might change** (you might add or remove items)

At LagosGrid: bus stops on a route are a perfect list — they're ordered, and routes can be updated.

```python
# ── list examples at LagosGrid ───────────────

bus_stops = ["Mile 2", "Alaba", "Festac", "Orile", "CMS", "TBS"]

print("All stops:    ", bus_stops)
print("First stop:   ", bus_stops[0])       # indexing starts at 0
print("Last stop:    ", bus_stops[-1])       # -1 = last element
print("Stops 1–3:   ", bus_stops[1:4])      # slicing
print("Total stops:  ", len(bus_stops))
print()

# ── Mutating the list ────────────
print("── Modifying the route ────")
bus_stops.append("Marina")               # add to end
print("After append: ", bus_stops)

bus_stops.insert(2, "Trade Fair")        # insert at index 2
print("After insert: ", bus_stops)

bus_stops.remove("Alaba")               # remove by value
print("After remove: ", bus_stops)

last_stop = bus_stops.pop()             # remove and return last item
print("Popped:       ", last_stop)
print("After pop:    ", bus_stops)
print()

```

```python
# ── Useful list operations ───────
fares = [350.50, 420.00, 280.00, 390.00, 175.50]

print("Fares:        ", fares)
print("Total fares:  ", sum(fares))
print("Highest fare: ", max(fares))
print("Lowest fare:  ", min(fares))
print("Sorted:       ", sorted(fares))
print()

# ── List of dicts (how real DE data arrives) ──
trip_log = [
    {"trip": "Morning", "bus_id": 1042, "fare": 350.50, "passengers": 45},
    {"trip": "Midday",  "bus_id": 1087, "fare": 280.00, "passengers": 28},
    {"trip": "Evening", "bus_id": 1042, "fare": 420.00, "passengers": 61},
]
print("Trip log has", len(trip_log), "entries")
print("Morning fare: ", trip_log[0]["fare"])
```

### 3.2  `tuple` — Ordered, Immutable Sequence

Use a `tuple` when:
- The **order matters**, AND
- The **data must never change** — it's a permanent record

At LagosGrid: GPS coordinates of a terminal are a perfect tuple.  
The CMS terminal's location is fixed. If it ever changes, something has gone very wrong.

```python
# ── tuple examples at LagosGrid ───────────────

# Terminal GPS coordinates — these are FIXED geographic locations
cms_terminal   = (6.4541, 3.3947)    # CMS Bus Terminal
tbs_terminal   = (6.4698, 3.3851)    # TBS Bus Terminal
mile2_terminal = (6.4802, 3.2883)    # Mile 2 Terminal

print("CMS Terminal:    ", cms_terminal)
print("Latitude:        ", cms_terminal[0])      # indexing works same as list
print("Longitude:       ", cms_terminal[1])
print()

# ── Unpacking ────────────────
lat, lon = cms_terminal
print(f"CMS is at lat={lat}, lon={lon}")

origin, destination = tbs_terminal, mile2_terminal
print(f"Trip: {origin} → {destination}")
print()

```

```python
# ── Tuples are IMMUTABLE — this will fail ─────────────────────────────────────
print("Trying to change a coordinate...")
try:
    cms_terminal[0] = 0.0    # this should fail
    print("Changed! (this should not print)")
except TypeError as e:
    print(f" Blocked correctly: {e}")
    print("  The tuple protected the data integrity.")
print()
```

```python
# ── Other tuple uses at LagosGrid ────
# Composite database key (route + date — uniquely identifies a trip)
trip_key   = ("R207", "2026-06-28")
print("Trip key:    ", trip_key)

# RGB colour for route map display
route_colour = (232, 96, 48)   # LagosGrid orange in RGB
print("Route colour:", route_colour)

# Multiple return values (functions implicitly return tuples)
def get_terminal_info(name, coords):
    return name, coords[0], coords[1]

name, lat, lon = get_terminal_info("CMS", cms_terminal)
print(f"{name} → ({lat}, {lon})")
```

### 3.3  `list` vs `tuple` — Which One When?

| | `list` | `tuple` |
|---|---|---|
| **Mutable** |  Yes |  No |
| **Memory** | More (dynamic sizing) | Less (fixed) |
| **Speed** | Slightly slower | Faster to iterate |
| **Use when** | Data can change over time | Data is a permanent record |
| **LagosGrid example** | Bus stops (routes change) | Terminal GPS (location is fixed) |

> **DE rule of thumb:** If the data represents a configuration or record that should never be modified after creation — use a `tuple`. Immutability is a data contract, not just a Python detail.

### 3.4  `dict` — Key-Value Store

Use a `dict` when:
- Data has **named fields** (like a database row or JSON record)
- You need to **look up values by name** (not by position)

At LagosGrid: every passenger record, every trip record, every API response is a `dict`.

```python
# ── dict examples at LagosGrid ───────────

passenger = {
    "id":     4821,
    "name":   "Amaka Osei",
    "age":    28,
    "route":  "Mile 2 - TBS Express",
    "fare":   350.50,
    "paid":   True,
    "stops":  ["CMS", "VI", "Lekki"],       # dict can contain lists
    "coords": (6.4541, 3.3947),             # dict can contain tuples
}

print("Passenger record:")
for key, value in passenger.items():
    print(f"  {key:8s}: {value}")
print()

# ── Accessing values ──────────────────────────────────────────────────────────
print("Name:         ", passenger["name"])
print("Fare:         ", passenger["fare"])
print("First stop:   ", passenger["stops"][0])      # list inside dict
print("Latitude:     ", passenger["coords"][0])     # tuple inside dict
print()

# ── Safe access with .get() ───────────────────────────────────────────────────
# Use .get() when a key might not exist — avoids KeyError
age      = passenger.get("age", "Unknown")          # returns 28
discount = passenger.get("discount_code", None)     # returns None (not an error)
print("Age:          ", age)
print("Discount:     ", discount)
print()

# ── Modifying a dict ──────────────────────────────────────────────────────────
passenger["status"]     = "CONFIRMED"         # add new key
passenger["fare"]       = 320.00              # update existing key
del passenger["age"]                          # remove a key

print("Updated keys: ", list(passenger.keys()))
print("Status:       ", passenger["status"])
print("New fare:     ₦", passenger["fare"])
print()

# ── Dict comprehension (DE pattern) ─────────
# Create a route-to-distance lookup table
routes = ["R101", "R205", "R207", "R310"]
distances = [8.2, 14.7, 22.1, 6.5]

route_map = {route: dist for route, dist in zip(routes, distances)}
print("Route distances:", route_map)
print("R207 distance:  ", route_map["R207"], "km")
```

### 3.5  `set` — Unique Values Only

Use a `set` when:
- You need **only unique values** (automatic deduplication)
- You need to do **set operations**: intersection, union, difference

At LagosGrid: unique route codes, unique terminals visited, unique passengers per day — all perfect for `set`.

```python
# ── set examples at LagosGrid ─────────────

# Amaka's trip log this week (with duplicates — she travels R207 every day)
amaka_trips = ["R207", "R101", "R207", "R310", "R207", "R101", "R205"]

# Convert to set — duplicates removed automatically
amaka_routes = set(amaka_trips)
print("Raw trip log:    ", amaka_trips)
print("Unique routes:   ", amaka_routes)
print("Unique count:    ", len(amaka_routes))
print()

# ── Building a set directly ───────────────────────────────────────────────────
morning_routes = {"R101", "R205", "R207"}
evening_routes = {"R101", "R310", "R207"}

print("Morning routes:  ", morning_routes)
print("Evening routes:  ", evening_routes)
print()


```

```python
# ── Set operations (critical DE skill) ───────────────────────────────────────
both_periods   = morning_routes & evening_routes    # intersection
either_period  = morning_routes | evening_routes    # union
morning_only   = morning_routes - evening_routes    # difference
evening_only   = evening_routes - morning_routes    # difference
not_in_common  = morning_routes ^ evening_routes    # symmetric difference

print("Both periods:    ", both_periods)     # routes that run morning AND evening
print("Either period:   ", either_period)    # all routes that run at any time
print("Morning only:    ", morning_only)     # routes that only run in the morning
print("Evening only:    ", evening_only)     # routes that only run in the evening
print("Not in common:   ", not_in_common)    # routes that run in ONE period only
print()

# ── Membership testing (sets are O(1) — instant, regardless of size) ──────────
print("Is R207 in morning routes?", "R207" in morning_routes)
print("Is R999 in any routes?    ", "R999" in either_period)
print()

# ── Frozen set — immutable set (like tuple for sets) ─────────────────────────
approved_routes = frozenset({"R101", "R205", "R207", "R310"})
print("Approved routes (frozen):", approved_routes)
# approved_routes.add("R999")  # - would raise AttributeError
```

### 4.1  `type()` and `isinstance()`

Two tools for inspecting types — you'll use these constantly when writing validation functions.

```python
# ── type() vs isinstance() ────────

fare = 350.50
name = "Amaka"
bus_id = 1042
active = True

# type() returns the exact type
print("type(fare):   ", type(fare))
print("type(name):   ", type(name))
print("type(bus_id): ", type(bus_id))
print("type(active): ", type(active))
print()

# isinstance() checks if a value is an instance of a type (or subclass)
print("isinstance(fare,   float) -", isinstance(fare,   float))   # True
print("isinstance(bus_id, int)   -", isinstance(bus_id, int))     # True
print("isinstance(name,   str)   -", isinstance(name,   str))     # True
print()


```

```python
# ── The bool trap ──────────
print("── Bool is a subclass of int ─────────────────────")
print("isinstance(True, int)    -", isinstance(True, int))        # True  - !!
print("isinstance(True, bool)   -", isinstance(True, bool))       # True
print("type(True) == bool       -", type(True) == bool)           # True  - safest
print("type(True) == int        -", type(True) == int)            # False  - correct
print()

# ── isinstance() with multiple types (tuple of types) ─────
def is_numeric(val):
    return isinstance(val, (int, float)) and not isinstance(val, bool)

print("is_numeric(350.50) →", is_numeric(350.50))   # True
print("is_numeric(45)     →", is_numeric(45))         # True
print("is_numeric('350')  →", is_numeric('350'))      # False
print("is_numeric(True)   →", is_numeric(True))       # False (excluded bool!)
```

### 4.2  Type Conversion (Casting)

Casting changes a value from one type to another.  

> **Real DE scenario:** Every JSON API response arrives with **all values as strings**.  
> Your first job in any ETL pipeline is to cast the raw strings into the correct types.

```python
# ── Casting functions ─────────────────────────────────────────────────────────

# int()
print("int('1042')     -", int('1042'))        # string → int
print("int(3.9)        -", int(3.9))           # float  → int (truncates!)
print("int(True)       -", int(True))          # bool   → int
print()

# float()
print("float('350.50') -", float('350.50'))    # string → float
print("float(45)       -", float(45))          # int    → float
print()

# str()
print("str(1042)       -", str(1042))          # int    → string
print("str(350.50)     -", str(350.50))        # float  → string
print("str(True)       -", str(True))          # bool   → string
print()

# bool()  ← CAREFUL HERE
print("── bool() casting — the dangerous one ────")
print("bool(0)         -", bool(0))            # → False
print("bool(1)         -", bool(1))            # → True
print("bool(42)        -", bool(42))           # → True  (any non-zero int)
print("bool('')        -", bool(''))           # → False (empty string)
print("bool('False')   -", bool('False'))      # → True  ← !! non-empty string!
print("bool('0')       -", bool('0'))          # → True  ← !! '0' is not 0
print()
print(" bool('False') is TRUE because 'False' is a non-empty string!")
print("   To convert a string boolean, compare it to 'True':")
print("   'False' == 'True'  -", 'False' == 'True')    # - False 
print("   'True'  == 'True'  -", 'True'  == 'True')    # - True  
```

---
### 4.3  Fixing a Real API Payload

This is exactly what you will do on Day 1 of a data engineering job.

```python
# ── Simulated API response from LagosGrid's legacy system ─
# JSON APIs return everything as strings — this is the raw payload:

import json

raw_json = json.dumps({
    "bus_id":    "1042",
    "fare":      "350.50",
    "is_ac":     "True",
    "count":     "45",
    "is_delayed":"False",
    "stops":     "CMS,VI,Lekki,Ajah",
    "latitude":  "6.5244",
    "longitude": "3.3792",
})

raw = json.loads(raw_json)   # parse JSON - Python dict (but all values are still strings!)

print("RAW payload (all strings):")
for k, v in raw.items():
    print(f"  {k:12s}: {repr(v):20s}  type={type(v).__name__}")

print()


```

```python
# ── Your DE job: write the cast_record() function ───
def cast_record(raw: dict) -> dict:
    '''
    Cast a raw string-typed LagosGrid API record into correct Python types.
    Returns a new dict with properly typed values.
    '''
    return {
        "bus_id":     int(raw["bus_id"]),
        "fare":       float(raw["fare"]),
        "is_ac":      raw["is_ac"] == "True",         # ← compare to string, don't use bool()
        "count":      int(raw["count"]),
        "is_delayed": raw["is_delayed"] == "True",    # ← same here
        "stops":      raw["stops"].split(","),         # ← string → list
        "latitude":   float(raw["latitude"]),
        "longitude":  float(raw["longitude"]),
    }

clean = cast_record(raw)

print("CLEAN record (correct types):")
for k, v in clean.items():
    print(f"  {k:12s}: {repr(v):30s}  type={type(v).__name__}")

print()
```

```python
# ── Verify the types are correct ──────────────────────────────────────────────
assert type(clean["bus_id"])     == int,   "bus_id must be int"
assert type(clean["fare"])       == float, "fare must be float"
assert type(clean["is_ac"])      == bool,  "is_ac must be bool"
assert type(clean["count"])      == int,   "count must be int"
assert type(clean["is_delayed"]) == bool,  "is_delayed must be bool"
assert type(clean["stops"])      == list,  "stops must be list"
assert type(clean["latitude"])   == float, "latitude must be float"
print(" All type assertions passed — record is clean and ready for the pipeline!")
print()

# ── Now arithmetic works correctly ────────────────────────────────────────────
discounted_fare = clean["fare"] * 0.9 if clean["is_ac"] else clean["fare"]
print(f"Original fare:    ₦{clean['fare']:.2f}")
print(f"AC discount (10%): ₦{clean['fare'] * 0.1:.2f}")
print(f"Final fare:       ₦{discounted_fare:.2f}")
print(f"Stop count:       {len(clean['stops'])}")
```

### Step 3 — Create the Route Schedule (list of dicts)

Build a schedule of 3 trips today — each trip is a dict, all trips collected into a list.

```python
# ── STEP 3: Build the route schedule ─────────────────────────────────────────

# Bus stop sequence — ordered list (we travel in this order)
bus_stops = ["Mile 2", "Alaba", "Trade Fair", "Orile", "CMS", "TBS"]

# Terminal GPS coordinates — fixed, so tuple
terminal_coords = (6.4541, 3.3947)    # CMS terminal

# Schedule of trips today — list of dicts
schedule = [
    {
        "trip_id":    "T001",
        "period":     "Morning",
        "bus_id":     1001,
        "route":      "R101",
        "fare":       320.00,
        "passengers": 52,
        "is_delayed": False,
    },
    {
        "trip_id":    "T002",
        "period":     "Midday",
        "bus_id":     1042,
        "route":      "R207",
        "fare":       350.50,
        "passengers": 28,
        "is_delayed": False,
    },
    {
        "trip_id":    "T003",
        "period":     "Evening",
        "bus_id":     1067,
        "route":      "R101",
        "fare":       420.00,
        "passengers": 61,
        "is_delayed": True,
    },
]

```

```python
# ── Verify Step 3 ─────────────────────────────────────────────────────────────
print("Step 3 — Route schedule created:")
print(f"  Bus stops ({len(bus_stops)}):  {bus_stops}")
print(f"  Terminal:            {terminal_coords}")
print(f"  Trips in schedule:   {len(schedule)}")
print()
for trip in schedule:
    status = " DELAYED" if trip["is_delayed"] else "✓ ON TIME"
    print(f"  {trip['period']:8s}  Bus {trip['bus_id']}  "
          f"Route {trip['route']}  ₦{trip['fare']:.2f}  "
          f"{trip['passengers']} pax  {status}")
print()

# ── Analytics on the schedule ─────────────────────────────────────────────────
total_revenue = sum(t["fare"] * t["passengers"] for t in schedule)
total_pax     = sum(t["passengers"] for t in schedule)
avg_fare      = sum(t["fare"] for t in schedule) / len(schedule)
delayed_count = sum(1 for t in schedule if t["is_delayed"])

print(f"  Total revenue:   {total_revenue:,.2f}")
print(f"  Total passengers: {total_pax}")
print(f"  Average fare:    {avg_fare:.2f}")
print(f"  Delayed trips:    {delayed_count}/{len(schedule)}")
```

---
### Step 4 — Extract Unique Routes with a Set

```python
# ── STEP 4: Deduplicate route codes using a set ──────────

# Extract all route codes from the schedule (will have duplicates)
all_route_codes = [trip["route"] for trip in schedule]
print("All route codes (with duplicates):", all_route_codes)

# Convert to set — duplicates removed automatically
unique_routes = set(all_route_codes)
print("Unique routes:                    ", unique_routes)
print("Unique count:                     ", len(unique_routes))
print()

# ── Additional set operations ───────
# Which routes are approved vs which we're running?
approved_routes = {"R101", "R205", "R207", "R310", "R415"}
running_routes  = unique_routes   # from the schedule

unapproved = running_routes - approved_routes
unscheduled = approved_routes - running_routes

print("Running routes:    ", running_routes)
print("Approved routes:   ", approved_routes)
print("Unapproved (risk): ", unapproved if unapproved else "None")
print("Not running today: ", unscheduled)
print()

# ── Format route list for output ─────────────────────────────────────────────
routes_display = ", ".join(sorted(unique_routes))
print("Routes today:", routes_display)
```
